```python

#funcion recoreo palabra
#obtiene prefijos
#regresa prefijos para contar despues
def recorrepalabra(palabra):
    i = 2
    # len() obtiene la longitud de la palabra
    contar=[]
    while i < len(palabra):                
        #print(palabra[0:i])  # Accede al carácter actual
        prefijo=(palabra[0:i])  # Accede al carácter actual
        contar.append(prefijo)
        i += 1             # Incrementa el índice
    return contar
        

        
#lista de palabras origen
frutas = ["manzana", "banana", "cereza", "naranja","naranja","naranja",  "naranja", "naranja", "bananaewewewew", "naranja"]
#lista de palabras para contar
listapalabrascontar=[]
#ordernar lista
frutas.sort()
#recorrer lista
for palabra in frutas:
    resultado=recorrepalabra(palabra)
    listapalabrascontar.extend(resultado)

listaconteo=[ ]
#obtener valores unicos para contar
unicos = list(set(listapalabrascontar))
#conteo de prefijo
for palabra in unicos:
    conteo = listapalabrascontar.count(palabra)
    #print(palabra) #    print(conteo)
    listaconteo+=[(palabra, conteo)]
    
    
 
#ordnar lista por mayor conteo a menor conteo
listaconteo.sort(key=lambda x: x[1], reverse=True)
#print(listaconteo)  



#resultado lista de prefijos y conteo

for v1, v2 in listaconteo:
    print ("palabra="+v1 + " Conteo="+str(v2))




```
