```python
import numpy as np
import pandas as pd
valor_x=np.linspace(0,20,10)
valor_x
```

```python
def elevar_cuadrado_mas_cinco(datos):
    """Eleva al cuadrado y suma 5 a cada elemento del array."""
    return (datos ** 2) + 5

valor_y=elevar_cuadrado_mas_cinco(valor_x)
valor_y
```

```python
figura=plt.figure( )
grafico1 =figura.add_axes([0,0,1,1])
grafico1.plot(valor_x, valor_y,color='r',  linestyle=':', linewidth=5 )
```

```python
# Importar la librería matplotlib para graficar
import matplotlib.pyplot as plt

# Definir la función graficar_linea
def graficar_linea(x, y):
    """
    Dada una lista de valores de x (horas de estudio) y una lista de valores de y (calificaciones),
    grafica una línea que muestra la relación entre ambos.
    
    Args:
    x (list): Lista de horas de estudio.
    y (list): Lista de calificaciones.
    """
    # Crear la gráfica
    figura=plt.figure( )
    grafico1 =figura.add_axes([0,0,1,1])
   
    
    # Configurar el título y las etiquetas de los ejes
    plt.title('grafico 1')
    plt.xlabel('etiquetas x')
    plt.ylabel('etiquetas y')
       
    # Mostrar la gráfica
    grafico1.plot(x, y,color='g'  )
   
        
# Datos proporcionados
horas_estudio = [1, 2, 3, 4, 5, 6, 7, 8]
calificaciones = [55, 60, 65, 70, 75, 80, 85, 90]

# Llamar a la función para graficar
graficar_linea(horas_estudio, calificaciones)

    
```

```python
import matplotlib.pyplot as plt

def graficar_temperaturas(dias, temperaturas):
    # Crear el gráfico de líneas    
 
    figura=plt.figure( )
    grafico1 =figura.add_axes([0,0,1,1])    
    # Configurar el título y las etiquetas de los ejes
    plt.title('temperaturas')
    plt.xlabel('dias ')
    plt.ylabel('grados C')
    
    # Mostrar la leyenda    

    # Mostrar la gráfica
    grafico1.plot(dias, temperaturas,color='blue', label= 'temperaturas', linestyle='--' ) 
    grafico1.legend(loc=0)
 
# Días de la semana
dias = ['Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes', 'Sábado', 'Domingo']
# Temperaturas correspondientes a cada día (en grados Celsius)
temperaturas = [22, 24, 23, 25, 26, 28, 27]
# Llamar a la función para graficar las temperaturas
graficar_temperaturas(dias, temperaturas)
```
