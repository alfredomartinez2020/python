```python
import numpy as np
from sklearn.svm import SVC

# Clase que representa una muestra de aire
class AirSample:
    def __init__(self, pm25, pm10, o3, no2, quality=None):
        self.pm25 = pm25
        self.pm10 = pm10
        self.o3 = o3
        self.no2 = no2
        self.quality = quality  # 0 = saludable, 1 = contaminado

    def to_vector(self):
        return [self.pm25, self.pm10, self.o3, self.no2]


# Generador de datos sintéticos
class AirDataGenerator:
    def __init__(self, num_samples=200):
        self.num_samples = num_samples

    def generate(self):
        np.random.seed(42)  # Para reproducibilidad
        samples = []
        
        for _ in range(self.num_samples):
            pm25 = np.random.uniform(5, 100)
            pm10 = np.random.uniform(10, 150)
            o3 = np.random.uniform(10, 100)
            no2 = np.random.uniform(5, 80)
            
            # Asignar calidad según las reglas
            if pm25 > 35 or pm10 > 50 or no2 > 40:
                quality = 1  # contaminado
            else:
                quality = 0  # saludable
            
            samples.append(AirSample(pm25, pm10, o3, no2, quality))
        
        return samples


# Clasificador con SVM
class AirQualityClassifier:
    def __init__(self):
        self.model = SVC()

    def fit(self, samples):
        # Convertir samples a vectores de características y etiquetas
        X = np.array([sample.to_vector() for sample in samples])
        y = np.array([sample.quality for sample in samples])
       
        print(X)
       
       
        self.model.fit(X, y)

    def predict(self, sample):
        # Convertir la muestra a vector y predecir
        X = np.array([sample.to_vector()])
        prediction = self.model.predict(X)
        return prediction[0]


# Ejemplo de uso
class AirQualityExample:
    def run(self):
        # Crear generador con 200 muestras
        generator = AirDataGenerator(num_samples=200)
        
        # Generar datos de entrenamiento
        training_data = generator.generate()
        
        # Crear y entrenar el clasificador
        classifier = AirQualityClassifier()
        classifier.fit(training_data)
        
        # Crear nueva muestra para predicción
        new_sample = AirSample(pm25=22, pm10=30, o3=50, no2=35)
        
        # Predecir calidad del aire
        prediction = classifier.predict(new_sample)
        
        # Mostrar resultados
        print("🌍 Muestra de aire:")
        print(f"PM2.5: {new_sample.pm25}, PM10: {new_sample.pm10}, O3: {new_sample.o3}, NO2: {new_sample.no2}")
        
        if prediction == 0:
            print("✅ Predicción de calidad: Saludable ✅")
        else:
            print("⚠️ Predicción de calidad: Contaminado ⚠️")


# Ejecutar ejemplo
example = AirQualityExample()
example.run()
```
