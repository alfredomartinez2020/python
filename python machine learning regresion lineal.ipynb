```python
from sklearn.linear_model import LinearRegression

# --- Clase Player ---
class Player:
    def __init__(self, name, avg_session_time, avg_actions_per_min, avg_kills_per_session, victories=0):
        self.name = name
        self.avg_session_time = avg_session_time
        self.avg_actions_per_min = avg_actions_per_min
        self.avg_kills_per_session = avg_kills_per_session
        self.victories = victories

    def to_features(self):
        return [self.avg_session_time, self.avg_actions_per_min, self.avg_kills_per_session]

# --- Clase PlayerDataset ---
class PlayerDataset:
    def __init__(self, players):
        self.players = players

    def get_feature_matrix(self):
        return [player.to_features() for player in self.players]

    def get_target_vector(self):
        return [player.victories for player in self.players]

# --- Clase VictoryPredictor (PARA COMPLETAR) ---
class VictoryPredictor:
    def __init__(self):
        # TODO: Crear el modelo LinearRegression
        self.modelo = LinearRegression()

    def train(self, dataset: PlayerDataset):
        # TODO: Obtener X e y del dataset y entrenar el modelo
        # Obtener X (características) e y (objetivo) del dataset
        X = dataset.get_feature_matrix()
        y = dataset.get_target_vector()        
        # Entrenar el modelo
        self.modelo.fit(X, y)
        print("Modelo entrenado con éxito.")     
        
        
        

    def predict(self, player: Player):
        # TODO: Predecir usando el modelo entrenado
        # Predecir usando el modelo entrenado
        # Convertimos las características del jugador en una matriz 2D
        features = np.array(player.to_features()).reshape(1, -1)
        prediction = self.modelo.predict(features)
        return prediction[0]
        
 

# --- Ejemplo de uso ---
# 1. Crear datos
p1 = Player("Alpha", 60, 100, 10, 50)
p2 = Player("Beta", 30, 50, 2, 10)
dataset = PlayerDataset([p1, p2])

# 2. Entrenar
predictor = VictoryPredictor()
predictor.train(dataset)

# 3. Predecir
nuevo_jugador = Player("Gana", 45, 80, 6)
prediccion = predictor.predict(nuevo_jugador)
print(f"Predicción de victorias para {nuevo_jugador.name}: {prediccion:.2f}")        
        
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np
class App:
    def __init__(self, name, downloads, rating, size_mb, reviews, revenue=None):
        self.name = name
        self.downloads = downloads  # en miles
        self.rating = rating        # de 1 a 5
        self.size_mb = size_mb
        self.reviews = reviews
        self.revenue = revenue      # en miles de dólares (puede ser None)

    def to_features(self):
        return [self.downloads, self.rating, self.size_mb, self.reviews]
        
        

# --- Clase RevenuePredictor (PARA COMPLETAR) ---
class RevenuePredictor:
    def __init__(self):
        # TODO: Crear el modelo LinearRegression
        self.modelo = LinearRegression()
        


    def fit(self, apps):
        # TODO: Filtrar apps con revenue conocido y entrenar el modelo
        # Filtrar apps con revenue conocido
        training_data = [app for app in apps if app.revenue is not None]        
        # Preparar X (features) y y (target/revenue)
        X = [app.to_features() for app in training_data]
        y = [app.revenue for app in training_data]        
        # Entrenar el modelo
        self.modelo.fit(X, y)        
        
        
        

    def predict(self, app):
        # TODO: Predecir ingresos de una nueva app
        features = np.array(app.to_features()).reshape(1, -1)
        return self.modelo.predict(features)[0]        
        
    
    
    


# Datos simulados de entrenamiento
training_apps = [
    App("TaskPro", 200, 4.2, 45.0, 1800, 120.0),
    App("MindSpark", 150, 4.5, 60.0, 2100, 135.0),
    App("WorkFlow", 300, 4.1, 55.0, 2500, 160.0),
    App("ZenTime", 120, 4.8, 40.0, 1700, 140.0),
    App("FocusApp", 180, 4.3, 52.0, 1900, 130.0),
    App("BoostApp", 220, 4.0, 48.0, 2300, 145.0),
]

# Creamos y entrenamos el predictor
predictor = RevenuePredictor()
predictor.fit(training_apps)

# Nueva app para predecir
new_app = App("FocusMaster", 250, 4.5, 50.0, 3000)
predicted_revenue = predictor.predict(new_app)

print(f"Ingresos estimados para {new_app.name}: ${predicted_revenue:.2f}K")




```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Clase que representa un registro de uso de un vehículo militar
class VehicleRecord:
    def __init__(self, hours_used, wear_level):
        self.hours_used = hours_used # Horas de uso
        self.wear_level = wear_level # Nivel de desgaste (0-100)
        
    def to_vector(self):
        return [self.hours_used]

# Clase para generar datos sintéticos
class VehicleDataGenerator:
    def __init__(self, num_samples=100):
        self.num_samples = num_samples
        
    def generate(self):
        np.random.seed(42) # Fijar semilla para reproducibilidad
        hours = np.random.uniform(50, 500, self.num_samples)
        # Simulamos un desgaste que aumenta con el uso (fórmula lineal + ruido)
        wear = 10 + 0.18 * hours + np.random.normal(0, 5, self.num_samples)
        wear = np.clip(wear, 0, 100) # Limitar a máximo 100
        data = [VehicleRecord(h, w) for h, w in zip(hours, wear)]
        return data

# --- Clase VehicleWearRegressor (COMPLETADA) ---
class VehicleWearRegressor:
    def __init__(self):
        # 1. Crear modelo LinearRegression
        self.model = LinearRegression()
        
    def fit(self, records):
        # 2. Preparar X e y y entrenar el modelo
        # X debe ser una matriz 2D (samples x features)
        X = np.array([r.hours_used for r in records]).reshape(-1, 1)
        # y es un vector 1D
        y = np.array([r.wear_level for r in records])
        
        self.model.fit(X, y)
        print("Modelo entrenado con éxito.")
        
    def predict(self, hours):
        # 3. Predecir el desgaste a partir de horas
        X_test = np.array([[hours]])
        prediction = self.model.predict(X_test)
        return prediction[0]
        
    def get_model(self):
        # 4. Devolver el modelo entrenado
        return self.model

# Clase principal para ejecutar el ejemplo
class VehicleWearPredictionExample:
    def run(self):
        # Generar datos
        generator = VehicleDataGenerator(100)
        records = generator.generate()
        
        # Entrenar regresor
        regressor = VehicleWearRegressor()
        regressor.fit(records)
        
        # Predecir el desgaste para un nuevo vehículo
        test_hours = 250
        prediction = regressor.predict(test_hours)
        
        print("-" * 30)
        print(f"⏱ Horas de uso estimadas: {test_hours}")
        print(f"⚙️ Nivel de desgaste estimado: {prediction:.2f}%")
        print("-" * 30)
        
        # Visualización
        df = pd.DataFrame({
            'Horas de uso': [r.hours_used for r in records],
            'Nivel de desgaste': [r.wear_level for r in records]
        })
        
        plt.figure(figsize=(8, 5))
        plt.scatter(df['Horas de uso'], df['Nivel de desgaste'], 
                    label='Datos reales', color='green', alpha=0.6)
        
        # Dibujar la línea de regresión
        x_line = np.linspace(40, 520, 100).reshape(-1, 1)
        y_line = regressor.get_model().predict(x_line)
        plt.plot(x_line, y_line, color='red', linewidth=2, label='Línea de regresión')
        
        # Marcar la predicción actual
        plt.axvline(test_hours, linestyle='--', color='gray', label=f'Vehículo nuevo ({test_hours}h)')
        plt.scatter(test_hours, prediction, color='blue', zorder=5, s=100)
        
        plt.title('📊 Relación entre horas de uso y desgaste (ML)')
        plt.xlabel('Horas de uso')
        plt.ylabel('Nivel de desgaste (%)')
        plt.legend()
        plt.grid(True, linestyle=':', alpha=0.6)
        plt.tight_layout()
        plt.show()

# Ejecutar el ejemplo
if __name__ == "__main__":
    example = VehicleWearPredictionExample()
    example.run()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Clase que representa un satélite
class Satellite:
    def __init__(self, duracion_mision, paneles_sol, carga_util, consumo_diario):
        self.duracion_mision = duracion_mision
        self.paneles_sol = paneles_sol
        self.carga_util = carga_util
        self.consumo_diario = consumo_diario

    def to_dict(self):
        return {
            "duracion_mision_dias": self.duracion_mision,
            "paneles_sol": self.paneles_sol,
            "carga_util": self.carga_util,
            "consumo_diario": self.consumo_diario,
        }

# Clase que genera el conjunto de datos sintético
class SatelliteDatasetGenerator:
    def __init__(self, n=300):
        self.n = n

    def generate(self):
        np.random.seed(42)
        duraciones = np.random.randint(100, 1000, self.n)
        paneles = np.random.uniform(10, 100, self.n)
        cargas = np.random.uniform(200, 2000, self.n)
        # Consumo: base + dependencia leve con duración y carga
        consumo = 5 + 0.01 * duraciones + 0.002 * cargas + np.random.normal(0, 1, self.n)

        satellites = []
        for d, p, c, e in zip(duraciones, paneles, cargas, consumo):
            satellites.append(Satellite(d, p, c, e))
        return satellites

# Clase que prepara los datos para análisis
class SatelliteDataProcessor:
    def __init__(self, satellites):
        self.df = pd.DataFrame([sat.to_dict() for sat in satellites])
        self.df["eficiencia_energia"] = self.df["consumo_diario"] / self.df["paneles_sol"]

    def get_dataframe(self):
        return self.df

# --- Clase a completar: regresión lineal ---
class EnergyConsumptionRegressor:
    def __init__(self):
        # 1. Crear modelo LinearRegression
        self.model = LinearRegression()

    def fit(self, X, y):
        # 2. Entrenar modelo
        self.model.fit(X, y)
        # 3. Devolver predicciones sobre los mismos datos X
        return self.model.predict(X)

    def evaluate(self, y_true, y_pred):
        # 4. Calcular R² usando r2_score
        r2 = r2_score(y_true, y_pred)
        print(f'R² (Coeficiente de determinación): {r2:.4f}')
        return r2

    def get_coefficients(self):
        # 5. Devolver coeficientes e intercepto
        return self.model.coef_, self.model.intercept_
        


# Clase que visualiza los datos y el modelo
class SatellitePlotter:
    def __init__(self, df, y_pred):
        self.df = df
        self.y_pred = y_pred

    def plot(self):
        plt.figure(figsize=(8, 6))
        scatter = plt.scatter(self.df["duracion_mision_dias"], self.df["consumo_diario"],
                              c=self.df["carga_util"], cmap="viridis", label="Datos reales", alpha=0.7)
        plt.plot(self.df["duracion_mision_dias"], self.y_pred, color="red", linewidth=2, label="Regresión lineal")
        plt.xlabel("Duración de la misión (días)")
        plt.ylabel("Consumo diario (kWh)")
        plt.title("🛰️ Consumo energético vs. duración de misión")
        cbar = plt.colorbar(scatter)
        cbar.set_label("Carga útil (kg)")
        plt.grid(True)
        plt.legend()
        plt.show()

# Clase principal que orquesta todo
class SatelliteAnalysisExample:
    def run(self):
        # 1. Generar datos
        generator = SatelliteDatasetGenerator()
        satellites = generator.generate()

        # 2. Procesar datos
        processor = SatelliteDataProcessor(satellites)
        df = processor.get_dataframe()

        # 3. Regresión lineal
        X = df[["duracion_mision_dias"]]
        y = df["consumo_diario"]
        regressor = EnergyConsumptionRegressor()
        y_pred = regressor.fit(X, y)

        # 4. Evaluar modelo
        r2 = regressor.evaluate(y, y_pred)
        coef, intercept = regressor.get_coefficients()
        print(f"📈 Modelo: y = {coef[0]:.4f} * x + {intercept:.2f}")
        print(f"🔍 R² del modelo: {r2:.4f}")

        # 5. Visualización
        plotter = SatellitePlotter(df, y_pred)
        plotter.plot()

# Ejecutar
if __name__ == "__main__":
    example = SatelliteAnalysisExample()
    example.run()


```
