```python
#de una frase
frase="hola esta es la frase para este ejercicio, favor de seguir instrucciones y realizar lo que se solicita de lo que"
#contar numero de palabras
frasesplit=frase.split()

#imprimir numero total de plabras
total = len(frasesplit)
print ("total " + str(total))
#imprimir numero de palabras repetidas
listasinrepetidas=set(frasesplit)#es un conjunto
print(listasinrepetidas)
print("palabras repetidas")
for p in listasinrepetidas:
    totalp=frasesplit.count(p)
    if (totalp>1):
        print(p+" "+str(totalp))
    
```
