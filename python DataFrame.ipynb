```python
import csv
import pandas as pd
# Existing DataFrame
df = pd.DataFrame({'Name': ['Alice', 'Bob'], 'Age': [25, 30]})
#otra forma de APPEND
df.loc[len(df.index)]=['luisalfredo', 45]
with open("/home/casa/nombres.csv") as file_obj:
    reader_obj=csv.DictReader(file_obj)
    print(reader_obj)
    for row in reader_obj:
        print(row)
        # Data to APPEND as a dictionary
        new_row = {'Name': row['nombre'], 'Age': row['edad']}
        # Append using concat
        df = pd.concat([df, pd.DataFrame([new_row])], ignore_index=True)
#imprimir DF
df
    
```
