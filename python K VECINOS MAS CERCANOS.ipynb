```python
dataframe =pd.read_csv('vehiculos.csv')
dataframe
```

```python
dataframe['vehicle_class'].unique()
```

```python
#variable objetivo Y que tipo de vehiculo
y=dataframe['vehicle_class']
y
```

```python
X=dataframe.drop('vehicle_class', axis =1)
X
```

```python
#objetyivo saber si el tipo de vehiculo de x
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test,y_train,y_test = train_test_split(X, y, test_size=0.3, random_state=45)
```

```python
X_train
```

```python
y_train
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn= KNeighborsClassifier(n_neighbors=1)
```

```python
#entrenar
knn.fit(X_train, y_train)
```

```python
#ahora predecir con modelo de train
predicciones=knn.predict(X_test)
predicciones
```

```python
#Una matriz de confusión (o confusion matrix) es una herramienta de evaluación para modelos de Machine Learning. 
#Resume el rendimiento comparando las predicciones de tu modelo contra los resultados reales
```

```python
from sklearn.metrics import classification_report, confusion_matrix
print(confusion_matrix(y_test, predicciones))
```

```python
#mostrar presicion del modelo
print(classification_report(y_test, predicciones))
```

```python
tasa_error = []
for i in range(1,30):
    knn = KNeighborsClassifier(n_neighbors=i)
    knn.fit(X_train, y_train)
    prediccion_i= knn.predict(X_test)
    tasa_error.append(np.mean(prediccion_i!=y_test))
```

```python
tasa_error
```

```python
valores=range(1,30)
plt.plot(valores, tasa_error, color='green', marker='o',markerfacecolor='red', markersize=8)
```
