```python
X=vinos.drop('Wine Type', axis =1)
X
```

```python
y=vinos['Wine Type']
y
```

```python
from sklearn.model_selection import train_test_split
#para entrenar
```

```python
X_train, X_test, y_train, y_test=train_test_split(X, y, test_size=0.3 )
```

```python
X_train
```

```python
from sklearn.tree import DecisionTreeClassifier #modelo de arbol de desicion
```

```python
arbol =DecisionTreeClassifier()
arbol.fit(X_train, y_train)
```

```python
predicciones=arbol.predict(X_test)
predicciones
```

```python
y_test
```

```python
from sklearn.metrics import classification_report, confusion_matrix
#confusion_matrix
#Es ideal para visualizar la distribución de los errores. 
#e permite detectar si tu modelo está confundiendo una clase específica con otra.

#classification_report
#Es un resumen estadístico que desglosa las métricas clave de evaluación para cada una de las clases de tu model
```

```python
print(classification_report(y_test, predicciones))
```

```python
print(confusion_matrix(y_test, predicciones))
```
