```python
import pandas as pd
import numpy as np
```

```python
#agrupacion de datos
dias = [ 'dia1','dia1','dia3','dia3','dia4'] 
vendedores = [ 'vendedor1','vendedor2','vendedor3','vendedor4','vendedor5'] 
ventas = [ 500,74,85, 9655, 6441] 
diccionario = { 'dias':dias  , 'vendedores': vendedores , 'ventas': ventas}

dataframe = pd.DataFrame(diccionario)

dataframe
```

```python
dataframe.groupby('dias')
```

```python
dataframe.groupby('dias').mean('ventas')
```

```python
dataframe.groupby('dias').sum('ventas')
```

```python
dataframe.groupby('dias').max('ventas')
```

```python
dataframe.groupby('dias').describe()
```
