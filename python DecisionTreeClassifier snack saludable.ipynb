```python
import numpy as np
from sklearn.tree import DecisionTreeClassifier

# Clase Snack
class Snack:
    def __init__(self, calories, sugar, protein, fat, fiber, is_healthy=None):
        self.calories = calories
        self.sugar = sugar
        self.protein = protein
        self.fat = fat
        self.fiber = fiber
        self.is_healthy = is_healthy

    def to_vector(self):
        return [self.calories, self.sugar, self.protein, self.fat, self.fiber]


# Clase para generar datos sintéticos
class SnackGenerator:
    def __init__(self, num_samples=100):
        self.num_samples = num_samples

    def generate(self):
        snacks = []
        for _ in range(self.num_samples):
            calories = np.random.uniform(50, 500)
            sugar    = np.random.uniform(0, 50)
            protein  = np.random.uniform(0, 30)
            fat      = np.random.uniform(0, 30)
            fiber    = np.random.uniform(0, 15)

            # Regla de salud
            is_healthy = int(
                calories < 200 and
                sugar    < 15  and
                fat      < 10  and
                (protein >= 5 or fiber >= 5)
            )

            snacks.append(Snack(calories, sugar, protein, fat, fiber, is_healthy))
        return snacks


# Clasificador usando árbol de decisión
class SnackClassifier:
    def __init__(self):
        self.model = DecisionTreeClassifier(random_state=42)

    def fit(self, snacks):
        X = [s.to_vector() for s in snacks]
        y = [s.is_healthy   for s in snacks]
        self.model.fit(X, y)

    def predict(self, snack):
        vector = [snack.to_vector()]          # shape (1, 5)
        return self.model.predict(vector)[0]  # devuelve 0 o 1


# Ejemplo de uso
class SnackRecommendationExample:
    def run(self):
        # 1. Generar datos sintéticos
        generator = SnackGenerator(num_samples=150)
        snacks = generator.generate()

        # 2. Entrenar el clasificador
        classifier = SnackClassifier()
        classifier.fit(snacks)

        # 3. Crear snack de prueba
        test_snack = Snack(
            calories=150,
            sugar=10,
            protein=6,
            fat=5,
            fiber=3
        )

        # 4. Predecir y mostrar resultado
        prediction = classifier.predict(test_snack)

        print("🔍 Snack Info:")
        print(f"   Calories: {test_snack.calories}, Sugar: {test_snack.sugar}g, "
              f"Protein: {test_snack.protein}g, Fat: {test_snack.fat}g, "
              f"Fiber: {test_snack.fiber}g")

        if prediction == 1:
            print("✅ Predicción: Este snack ES saludable.")
        else:
            print("❌ Predicción: Este snack NO es saludable.")


# Ejecutar ejemplo
example = SnackRecommendationExample()
example.run()
```
