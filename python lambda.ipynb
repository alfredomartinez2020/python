```python
#OPERACION NORMAL
def multiplicar(n):
    return n*22


multiplicar(5)


lista=[1,2,3,4,5]

lista2=list(map(multiplicar, lista))
lista2
```

```python
#OPERACION LAMDA O ANONIMA LO MISMO PERO CON LAMBDA
#parametro numero
#sobre cada elemento de lista 
#multiplica por 22
lista=[1,2,3,4,5]

lista2=list(map(lambda numero:numero * 22, lista))
lista2
    
```
