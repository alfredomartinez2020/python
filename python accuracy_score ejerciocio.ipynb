```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
import numpy as np

class ProPlayerClassifier:
    def __init__(self):
        self.model = SVC(kernel='rbf', random_state=42)
        
    def train(self, X, y):
        """Entrena el modelo SVM con los datos proporcionados"""
        self.model.fit(X, y)
        return self
        
    def predict(self, player_stats):
        """Predice si un jugador es profesional (1) o casual (0)"""
        # Asegurar que los datos tengan la forma correcta
        if len(player_stats.shape) == 1:
            player_stats = player_stats.reshape(1, -1)
        return self.model.predict(player_stats)
        
    def evaluate(self, X_test, y_test):
        """Evalúa el rendimiento del modelo con precisión"""
        predictions = self.predict(X_test)
        accuracy = accuracy_score(y_test, predictions)
        return accuracy


class GameSimulator:
    def __init__(self):
        # Generar datos simulados para entrenamiento y prueba
        np.random.seed(42)
        
        # Crear 50 jugadores profesionales (etiqueta 1)
        n_pro = 50
        pro_players = np.random.uniform(0.7, 1.0, (n_pro, 5))
        
        # Crear 50 jugadores casuales (etiqueta 0)
        n_casual = 50
        casual_players = np.random.uniform(0.0, 0.6, (n_casual, 5))
        
        # Combinar todos los datos
        self.X = np.vstack([pro_players, casual_players])
        self.y = np.array([1]*n_pro + [0]*n_casual)
        
        # Mezclar los datos
        shuffle_idx = np.random.permutation(len(self.y))
        self.X = self.X[shuffle_idx]
        self.y = self.y[shuffle_idx]
        
        # Dividir en entrenamiento (80%) y prueba (20%)
        split_idx = int(0.8 * len(self.y))
        self.X_train = self.X[:split_idx]
        self.y_train = self.y[:split_idx]
        self.X_test = self.X[split_idx:]
        self.y_test = self.y[split_idx:]
        
    def run(self):
        """Ejecuta la simulación completa"""
        # 1. Crear y entrenar el clasificador
        classifier = ProPlayerClassifier()
        classifier.train(self.X_train, self.y_train)
        
        # 2. Evaluar el modelo
        accuracy = classifier.evaluate(self.X_test, self.y_test)
        
        # 3. Crear jugadores de ejemplo para predicciones
        example_players = np.array([
            [0.95, 0.92, 0.88, 0.91, 0.89],  # Jugador profesional
            [0.35, 0.42, 0.28, 0.31, 0.39]   # Jugador casual
        ])
        
        # 4. Realizar predicciones
        predictions = classifier.predict(example_players)
        
        # 5. Mostrar resultados
        print("Jugador profesional:")
        print(f"Precisión del modelo: {accuracy:.2f}")
        
        # Mostrar predicciones para los jugadores de ejemplo
        print("\nPredicciones para jugadores de ejemplo:")
        for i, player in enumerate(example_players):
            status = "Profesional (1)" if predictions[i] == 1 else "Casual (0)"
            print(f"Jugador {i+1}: {player} → {status}")


# Ejecutar la simulación
simulator = GameSimulator()
simulator.run()
```

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
import numpy as np

class ProPlayerClassifier:
    def __init__(self):
        self.model = SVC(kernel='rbf', random_state=42)
        
    def train(self, X, y):
        """Entrena el modelo SVM con los datos proporcionados"""
        self.model.fit(X, y)
        return self
        
    def predict(self, player_stats):
        """Predice si un jugador es profesional (1) o casual (0)"""
        # Convertir a array numpy si es una lista
        if isinstance(player_stats, list):
            player_stats = np.array(player_stats)
        
        # Si es un solo jugador (vector 1D), convertirlo a matriz 2D
        if player_stats.ndim == 1:
            player_stats = [player_stats]
            
        return self.model.predict(player_stats)
        
    def evaluate(self, X_test, y_test):
        """Evalúa el rendimiento del modelo con precisión"""
        predictions = self.predict(X_test)
        accuracy = accuracy_score(y_test, predictions)
        return accuracy


class GameSimulator:
    def __init__(self):
        # Generar datos simulados para entrenamiento y prueba
        np.random.seed(42)
        
        # Crear 50 jugadores profesionales (etiqueta 1)
        n_pro = 50
        pro_players = []
        for _ in range(n_pro):
            pro_players.append([
                np.random.uniform(0.7, 1.0),  # partidas ganadas
                np.random.uniform(0.7, 1.0),  # horas jugadas
                np.random.uniform(0.7, 1.0),  # precisión
                np.random.uniform(0.7, 1.0),  # velocidad de reacción
                np.random.uniform(0.7, 1.0)   # estrategia
            ])
        
        # Crear 50 jugadores casuales (etiqueta 0)
        n_casual = 50
        casual_players = []
        for _ in range(n_casual):
            casual_players.append([
                np.random.uniform(0.0, 0.6),  # partidas ganadas
                np.random.uniform(0.0, 0.6),  # horas jugadas
                np.random.uniform(0.0, 0.6),  # precisión
                np.random.uniform(0.0, 0.6),  # velocidad de reacción
                np.random.uniform(0.0, 0.6)   # estrategia
            ])
        
        # Combinar todos los datos
        self.X = pro_players + casual_players
        self.y = [1]*n_pro + [0]*n_casual
        
        # Mezclar los datos
        indices = list(range(len(self.y)))
        np.random.shuffle(indices)
        self.X = [self.X[i] for i in indices]
        self.y = [self.y[i] for i in indices]
        
        # Convertir a arrays numpy
        self.X = np.array(self.X)
        self.y = np.array(self.y)
        
        # Dividir en entrenamiento (80%) y prueba (20%)
        split_idx = int(0.8 * len(self.y))
        self.X_train = self.X[:split_idx]
        self.y_train = self.y[:split_idx]
        self.X_test = self.X[split_idx:]
        self.y_test = self.y[split_idx:]
        
    def run(self):
        """Ejecuta la simulación completa"""
        # 1. Crear y entrenar el clasificador
        classifier = ProPlayerClassifier()
        classifier.train(self.X_train, self.y_train)
        
        # 2. Evaluar el modelo
        accuracy = classifier.evaluate(self.X_test, self.y_test)
        print('accuracy')
        print(accuracy)
        # 3. Crear jugadores de ejemplo para predicciones
        example_players = [
            [0.95, 0.92, 0.88, 0.91, 0.89],  # Jugador profesional
            [0.35, 0.42, 0.28, 0.31, 0.39]   # Jugador casual
        ]
        
        # 4. Realizar predicciones
        predictions = classifier.predict(example_players)
        
        # 5. Mostrar resultados
        print("Jugador profesional:")
        print(f"Precisión del modelo: {accuracy:.2f}")
        
        # Mostrar predicciones para los jugadores de ejemplo
        print("\nPredicciones para jugadores de ejemplo:")
        for i, player in enumerate(example_players):
            status = "Profesional (1)" if predictions[i] == 1 else "Casual (0)"
            print(f"Jugador {i+1}: {player} → {status}")


# Ejecutar la simulación
simulator = GameSimulator()
simulator.run()
```
