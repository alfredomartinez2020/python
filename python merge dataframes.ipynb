```python
##MERGE DATAFRAMES

import numpy as np
import pandas as pd


diccionario1 = {'c':[1,2,3], 'd':[4,5,6], 'clave':['c1','c2','c3']}
dataframe1 = pd.DataFrame(diccionario1)





diccionario2 = {'f':[1,2,3], 'e':[44,54,44], 'clave':['c1','c2','c3']}
dataframe2 = pd.DataFrame(diccionario2)

 
```

```python
dataframe1
```

```python
dataframe2
```

```python
pd.merge(dataframe1, dataframe2, on= 'clave')
```
