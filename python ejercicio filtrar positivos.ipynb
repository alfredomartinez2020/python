```python
#  de lista numero POSITIVOS Y NEGATIVOS
# crear funcion positivo 
# devuelve si es positivo o negativo
# crear lista 2 con filter solo positivos
#
#



lista=[1,-2,3,4,-5,6,7,8,-9,0,0,8,6,5,4,3,2,1,5,4,23]

def positivo(n):
    if(n>0):
        return ("positivo "+format(n))

for n in lista:
    positivo(n)


listapositivos=list(filter(positivo, lista))

listapositivos


 


```

```python
#  de lista numero POSITIVOS Y NEGATIVOS
# crear funcion positivo 
# devuelve si es positivo o negativo
# crear lista 2 con filter solo positivos
#
#



lista=[1,-2,3,4,-5,6,7,8,-9,0,0,8,6,5,4,3,2,1,5,4,23]

def positivo(n):
    if(n>0):
        return ("positivo "+format(n))

for n in lista:
    positivo(n)


 



listapositivos=list(filter(lambda n : n>0 , lista))

listapositivos
```
