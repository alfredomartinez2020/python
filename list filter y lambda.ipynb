```python
def pares(n):
    if (n % 2==0):
        return n
    else:
        return "impar"

def paresvalida(n):
    if (n % 2==0):
        return n
       
        
pares(4)
pares(3)

lista=[1,2,3,4,5,6,7,8,9,0]

lista_pares=list(filter(paresvalida, lista))
lista_pares
        
```
