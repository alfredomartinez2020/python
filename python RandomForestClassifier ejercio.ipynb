```python
import numpy as np
from sklearn.ensemble import RandomForestClassifier


class LanguagePredictor:
    def __init__(self):
        self.model = RandomForestClassifier()
        self.label_map = {
            0: "Python",
            1: "JavaScript",
            2: "Java",
            3: "C++"
        }

    def train(self, X: np.ndarray, y: np.ndarray):
        self.model.fit(X, y)

    def predict(self, features: np.ndarray) -> str:
        prediction = self.model.predict(features.reshape(1, -1))
        return self.label_map[prediction[0]]


def generate_dataset(n_samples=100, seed=42):
    rng = np.random.default_rng(seed)

    velocidad    = rng.random(n_samples)
    mantenimiento = rng.random(n_samples)
    libs         = rng.random(n_samples)
    tipo_app     = rng.integers(0, 3, n_samples)   # 0, 1 o 2
    rendimiento  = rng.random(n_samples)

    X = np.column_stack([velocidad, mantenimiento, libs, tipo_app, rendimiento])

    y = np.where(
        (rendimiento > 0.8) & (tipo_app == 2), 3,          # C++
        np.where(
            (mantenimiento > 0.7) & (tipo_app == 1), 2,    # Java
            np.where(
                (libs > 0.6) & (tipo_app == 0), 0,         # Python
                1                                           # JavaScript
            )
        )
    )

    return X, y


# Generar datos y entrenar
X, y = generate_dataset()
predictor = LanguagePredictor()
predictor.train(X, y)

# Crear un proyecto nuevo
new_project = np.array([0.7, 0.9, 0.5, 1, 0.6])  # Características del proyecto

# Predecir lenguaje ideal
pred = predictor.predict(new_project)
print(f"Lenguaje recomendado para el nuevo proyecto: {pred}")
```
